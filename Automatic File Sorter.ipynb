```python
import os, shutil
```

```python
path=r'C:/Users/user/Desktop/pythonn/Pyt_project_file_sorter/'
```

```python
#firlst let's check what kind of files we have
file_name=os.listdir(path)
print(file_name)
```

```python
#now let's create folder names
folder_names=['xlsx files', 'image files','pdf files', 'docs files']

for loop in range(0,4):
    if not os.path.exists(path+folder_names[loop]):
        print(path+folder_names[loop])
        os.makedirs(path+folder_names[loop])
```

```python
#Let's move files to the correspondence folder
for file in file_name:
    if '.docx' in file and not os.path.exists(path+'docs files/'+file):
        shutil.move(path+file,path+'docs files/'+file)
    elif '.xlsx' in file and not os.path.exists(path+'xlsx files/'+file):
        shutil.move(path+file,path+'xlsx files/'+file)
    elif '.jpg' in file and not os.path.exists(path+'image files/'+file):
        shutil.move(path+file,path+'image files/'+file) 
    elif '.pdf' in file and not os.path.exists(path+'pdf files/'+file):
        shutil.move(path+file,path+'pdf files/'+file) 
    else:
        print('there are files in this pasth that were not moved')
```
